```python
import pandas as pd
import re
import numpy as np


# funkcja czyszcząca tekst
def clean_whitespace(text):
    """
    Zamienia wszystkie znaki białe (new line, tab, wiele spacji) 
    na pojedynczą spację i usuwa spacje z początku/końca.
    """
    if isinstance(text, str):
        #### r'\s+' w wyrażeniach regularnych oznacza "jeden lub więcej znaków białych"
        #### (w tym spacje, entery \n, tabulacje \t)
        return re.sub(r'\s+', ' ', text).strip()
    return text


# wczytanie pliku
print("Wczytywanie pliku jstor_articles.csv...")
try:
    df = pd.read_csv("jstor_articles.csv")
    print(f"Wczytano {len(df):,} wierszy.")
except Exception as e:
    print(f"Błąd wczytywania: {e}")


# czyszczenie kolumn tekstowych #### wybiera kolumny które są typu 'object'
text_columns = ['keyword', 'title', 'abstract', 'creator', 'date', 'journal', 'doi', 'language']

print("Czyszczenie tekstów (usuwanie enterów i zbędnych spacji)...")

for col in text_columns:
    if col in df.columns:
        # wypełnianie pustych wartości (NaN) pustym stringiem, żeby funkcja działała a potem aplikacja czyszczenia
        df[col] = df[col].fillna("").astype(str).apply(clean_whitespace)


# dodatkowe sprawdzenie i usunięcie duplikatów
initial_count = len(df)
df = df.drop_duplicates()
final_count = len(df)
if initial_count > final_count:
    print(f"Usunięto {initial_count - final_count:,} zduplikowanych wierszy.")
```

```python
# podgląd danych po naprawie
print("\nPrzykładowe dane po naprawie:")
print(df[['title', 'creator']].head(10))
```

```python
# zapis do nowego pliku
output_file = "jstor_articles_fixed.csv"

df.to_csv(output_file, index=False, encoding='utf-8')

print(f"\nZapisanie do {output_file}...")
```
