```python
import pandas as pd

## wczytanie
input_file = "pubmed_articles.csv"
output_file = "pubmed_articles_fixed.csv"

df = pd.read_csv(input_file)
```

```python
# zlicza wiersze z 'No authors' w kolumnie Authors
no_authors = df['Authors'].value_counts()['No authors']
print(f"Liczba wierszy z 'No authors': {no_authors}")

# zastępuje 'No authors' pustym stringiem w kolumnie Authors
df['Authors'] = df['Authors'].replace(['No authors'], '')
```

```python
# zlicza wiersze z 'Unknown' w kolumnie Year
unknown_years = df['Year'].value_counts()['Unknown']
print(f"Liczba wierszy z 'Unknown' year: {unknown_years}")

# zastępuje 'Unknown' pustym stringiem w kolumnie Year
df['Year'] = df['Year'].replace(['Unknown'], '')
```

```python
# zlicza wiersze z 'No abstract' w kolumnie Abstract
no_abstracts = df['Abstract'].value_counts()['No abstract']
print(f"Liczba wierszy z 'No abstract': {no_abstracts}")

# zastępuje 'No abstract' pustym stringiem w kolumnie Abstract
df['Abstract'] = df['Abstract'].replace(['No abstract'], '')
```

```python
## zapis
print(f"Zapis do pliku {output_file}...")
df.to_csv(output_file, index=False, encoding='utf-8')
```
